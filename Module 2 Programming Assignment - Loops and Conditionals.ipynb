```python
secret = 4
guess = 5
if guess < secret:
    print('too low')
elif guess > secret:
    print('too high')
elif guess == secret:
    print('just right')
```

```python
small = False
green = False
if green and small:
    print('pea')
elif small and not green:
    print('cherry')
elif not small and green:
    print('watermelon')
elif not small and not green:
    print('pumpkin')
else:
    print('Selection has not yet been coded')
```

```python
guess_me = 7
number = 1
while True:
    if number < guess_me:
        print('too low')
    elif number == guess_me:
        print('found it!')
        break
    elif number > guess_me:
        print('oops')
        break
    number += 1
```

```python
guess_me = 5
for number in range(10):
    if number < guess_me:
        print('too low')
    elif number == guess_me:
        print('found it!')
        break
    elif number > guess_me:
        print('oops')
        break
```
