```python
import sqlite3

conn = sqlite3.connect('books.db')

curs = conn.cursor()
curs.execute('''
    CREATE TABLE books(title TEXT,
    author TEXT,
    year INTEGER)
''')
print("books table corrected")
curs.close()
conn.close()
```

```python
import csv
import sqlite3

conn = sqlite3.connect('books.db')
curs = conn.cursor()

with open('books2.csv', 'rt') as fin:
    reader = csv.DictReader(fin)

    for row in reader:
        curs.execute(
            '''
            INSERT INTO BOOKS VALUES (?, ?, ?)
            ''',
            (row['title'], row['author'], int(row['year']))
        )
    conn.commit()

    curs.execute('SELECT * FROM books')
    print(curs.fetchall())
    curs.close()
    conn.close()
```

```python
from sqlalchemy import create_engine, text

conn = create_engine('sqlite:///books.db')
rows = conn.connect().execute(
    text('SELECT title FROM books ORDER BY title')
)
for row in rows:
    print(row[0])
```
